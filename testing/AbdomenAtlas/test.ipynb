```python
import pandas as pd

# 1. Load your CSV file
csv_path = "data/AbdomenAtlas3.0MiniWithMeta.csv"  # <--- Replace with your CSV file name
df = pd.read_csv(csv_path)

# 2. Map your column names (update these string names if they differ in your CSV)
LIVER_COL = "number of liver lesion instances"       # e.g., 'liver_lesion_instances'
PANCREAS_COL = "number of pancreatic lesion instances" # e.g., 'pancreatic_lesion_instances'
KIDNEY_COL = "number of kidney lesion instances"     # e.g., 'kidney_lesion_instances'

# Define the pairs and triplet to inspect
combinations = {
    "Liver + Pancreas": [LIVER_COL, PANCREAS_COL],
    "Liver + Kidney": [LIVER_COL, KIDNEY_COL],
    "Pancreas + Kidney": [PANCREAS_COL, KIDNEY_COL],
    "Liver + Pancreas + Kidney (Triplet)": [LIVER_COL, PANCREAS_COL, KIDNEY_COL],
}

print("=" * 60)
print(f"{'CO-OCCURRENCE ANALYSIS':^60}")
print("=" * 60)

for combo_name, cols in combinations.items():
    # Filter rows where EVERY column in the pair/triplet is > 0
    co_occurring_mask = (df[cols] > 0).all(axis=1)
    matching_rows = df[co_occurring_mask]
    
    count = len(matching_rows)
    percentage = (count / len(df)) * 100 if len(df) > 0 else 0
    
    print(f"\n🔹 {combo_name}")
    print(f"   Co-occurring Cases (> 0 each): {count} / {len(df)} ({percentage:.2f}%)")
    
    if count > 0:
        print("   Average Lesion Counts (in co-occurring cases):")
        for col in cols:
            avg_count = matching_rows[col].mean()
            print(f"     - {col}: {avg_count:.2f} lesions / case")
    else:
        print("   Average Lesion Counts: N/A (no cases found)")

print("\n" + "=" * 60)
```

```python
import pandas as pd

# 1. Load your CSV file
csv_path = "data/AbdomenAtlas3.0MiniWithMeta.csv"
df = pd.read_csv(csv_path)

# 2. Define target columns
target_columns = [
    "number of liver lesion instances",
    "number of pancreatic lesion instances",
    "number of kidney lesion instances"
]

print("=" * 60)
print(f"{'INDIVIDUAL LESION ANALYSIS (EXCLUDING 0s)':^60}")
print("=" * 60)

for col in target_columns:
    # Filter rows where lesion count is 1 or higher (excluding 0)
    positive_cases = df[df[col] >= 1]
    
    count = len(positive_cases)
    total_rows = len(df)
    percentage = (count / total_rows * 100) if total_rows > 0 else 0
    
    print(f"\n🔹 {col}")
    print(f"   Positive Cases (>= 1): {count} / {total_rows} ({percentage:.2f}%)")
    
    if count > 0:
        avg_lesions = positive_cases[col].mean()
        print(f"   Average Lesions (in positive cases): {avg_lesions:.2f} lesions / case")
    else:
        print("   Average Lesions: N/A (no positive cases)")

print("\n" + "=" * 60)

# Optional: Total scans that have AT LEAST ONE lesion in any of these 3 organs
any_lesion_mask = (df[target_columns] >= 1).any(axis=1)
total_with_any_lesion = any_lesion_mask.sum()
print(f"Total scans with AT LEAST ONE lesion across these organs: "
      f"{total_with_any_lesion} / {len(df)} ({total_with_any_lesion / len(df) * 100:.2f}%)")
print("=" * 60)
```

```python
import pandas as pd
import numpy as np

# 1. Load CSV file
csv_path = "data/AbdomenAtlas3.0MiniWithMeta.csv"
df = pd.read_csv(csv_path)

AGE_COL = "age"  # Update if the column name differs (e.g., 'Age', 'patient_age')

# 2. Clean/Convert 'age' column to numeric (handling any missing or non-numeric values)
age_series = pd.to_numeric(df[AGE_COL], errors='coerce').dropna()

# 3. Basic Summary Statistics
total_rows = len(df)
valid_count = age_series.count()
missing_count = total_rows - valid_count
min_val = age_series.min()
max_val = age_series.max()
mean_val = age_series.mean()
median_val = age_series.median()
std_val = age_series.std()

print("=" * 50)
print(f"{'AGE COLUMN SUMMARY STATISTICS':^50}")
print("=" * 50)
print(f"Total Rows in Dataset : {total_rows}")
print(f"Valid Age Records     : {valid_count}")
print(f"Missing / NaN Records : {missing_count}")
print("-" * 50)
print(f"Minimum Age           : {min_val:.1f}")
print(f"Maximum Age           : {max_val:.1f}")
print(f"Mean Age              : {mean_val:.2f} years")
print(f"Median Age            : {median_val:.1f} years")
print(f"Standard Deviation    : {std_val:.2f}")
print("=" * 50)

# 4. Age Distribution Breakdown by Bins (Decades)
bins = [0, 18, 30, 40, 50, 60, 70, 80, 120]
labels = ["<18", "18-29", "30-39", "40-49", "50-59", "60-69", "70-79", "80+"]

age_binned = pd.cut(age_series, bins=bins, labels=labels, right=False)
dist_counts = age_binned.value_counts(sort=False)

print(f"\n{'AGE DISTRIBUTION BREAKDOWN':^50}")
print("-" * 50)
print(f"{'Age Bracket':<15} | {'Count':<10} | {'Percentage':<10}")
print("-" * 50)
for label, count in dist_counts.items():
    pct = (count / valid_count * 100) if valid_count > 0 else 0
    bar = "█" * int(pct // 2)  # Text-based bar visual
    print(f"{label:<15} | {count:<10} | {pct:>5.1f}%  {bar}")

print("=" * 50)
```

```python
import pandas as pd

# 1. Load your CSV file
df = pd.read_csv('data/AbdomenAtlas3.0MiniWithMeta.csv')

# 2. Define the attenuation columns for all 3 organs
attenuation_cols = {
    'Pancreas': 'largest pancreatic lesion attenuation (hyperattenuating, isoattenuating, hypoattenuating)',
    'Liver': 'largest liver lesion attenuation (hyperattenuating, isoattenuating, hypoattenuating)',
    'Kidney': 'largest kidney lesion attenuation (hyperattenuating, isoattenuating, hypoattenuating)'
}

# Dynamic fallback check in case exact column names differ slightly
matched_cols = {}
for organ, col in attenuation_cols.items():
    if col in df.columns:
        matched_cols[organ] = col
    else:
        # Search for columns containing organ name + attenuation
        found = [c for c in df.columns if organ.lower() in c.lower() and 'attenuation' in c.lower()]
        if found:
            matched_cols[organ] = found[0]
        else:
            print(f"⚠️ Warning: Could not find attenuation column for {organ}")

# 3. Print Value Counts for Each Organ
print("=" * 70)
print(f"{'LESION ATTENUATION VALUE COUNTS':^70}")
print("=" * 70)

for organ, col_name in matched_cols.items():
    print(f"\n🔹 {organ.upper()} (Column: '{col_name}')")
    print("-" * 70)
    
    # Get counts including NaN/missing
    counts = df[col_name].value_counts(dropna=False)
    percentages = df[col_name].value_counts(dropna=False, normalize=True) * 100
    
    summary = pd.DataFrame({
        'Count': counts,
        'Percentage (%)': percentages.round(2)
    })
    
    print(summary.to_string())

print("\n" + "=" * 70)
```
