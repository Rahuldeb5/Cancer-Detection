```python
import pandas as pd

# Load your CSV file
csv_file_path = "data/zero_shot_findings_disease_cls.csv"  # Replace with your actual file path
df = pd.read_csv(csv_file_path)

target_columns = ["pancreatic_atrophy", "biliary_ductal_dilation", "hydronephrosis", "surgically_absent_gallbladder"]

# Check counts
results = []
for col in target_columns:
    if col in df.columns:
        counts = df[col].value_counts()
        results.append({
            "Finding": col,
            "0s (Negative)": counts.get(0, 0),
            "1s (Positive)": counts.get(1, 0),
            "-1s (Uncertain)": counts.get(-1, 0),
            "Total Valid (0 & 1)": counts.get(0, 0) + counts.get(1, 0),
            "Total Rows": len(df[col])
        })
    else:
        print(f"Warning: Column '{col}' not found in CSV!")

# Display as a clean DataFrame
summary_df = pd.DataFrame(results)
print(summary_df.to_string(index=False))
```
